### Valores atípicos

Se utiliza el método del rango intercuartílico (IQR) para identificar si hay valores atípicos en las variables del IPI en el dataset.

Primero se calcula el Q1 y el Q3 para calcular el rango intercuartílico y obtener el límite inferior y el límite superior del mismo para todas las columnas numéricas del dataset

Así, todo valor que quede por debajo o por arriba de los límites se considera potencialmente atípico según este método.

```python
Q1 = df.quantile(0.25, numeric_only=True)
Q3 = df.quantile(0.75, numeric_only=True)

IQR = Q3 - Q1

limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR
```

Después, se crea una tabla con la cantidad de valores atípicos por variable.

```python
df_numerico = df.select_dtypes(include="number")

atipicos = (
    (df_numerico < limite_inferior) |
    (df_numerico > limite_superior)
).sum()

atipicos
```

Estos resultados indican que para las variables **IPI general**, **IPI general desestacionalizado**, IPI del sector **prendas_vestir_cuero_calzado**, IPI del sector **madera_papel_edicion_impresion**, IPI del sector **refinacion_petroleo**, IPI del sector **productos_minerales_no_metalicos**, IPI del sector **industrias_metalicas_basicas**, IPI del sector **productos_metal**,, IPI del sector **otros_equipos_aparatos_instrumentos**, IPI del sector **vehiculos_automotores_autopartes**, IPI del sector **otro_equipo_transporte** e IPI del sector **muebles_colchones_otras**, se encontraron valores potencialmente atípicos desde un punto de vista estadístico. Esto no quiere esos valores sean datos incorrectos, sino que pueden representar cambios más abruptos en la actividad industrial, y se analizará posteriormente si esos valores corresponden a errores en los datos o a variaciones reales de la actividad industrial. Es decir, simplemente indican dónde se debe prestar más atención en un análisis más profundo.

#### Serie de tiempo 

```python
plt.figure(figsize=(12, 5))

plt.plot(df["fecha"], df["ipi_general"])

plt.xlabel("Fecha")
plt.ylabel("Índice")
plt.title("Evolución temporal del IPI general")

plt.show()
```

La serie temporal del IPI general muestra una evolución variable durante el período analizado. Entre 2016 y 2020 se observa una tendencia general, aunque acompañada por sucesivos picos y descensos que reflejan variaciones en los niveles del índice a lo largo de los meses. En 2020 se produce una caída abrupta, que constituye uno de los cambios más marcados de toda la serie.

A partir de 2021 se observa una recuperación y nuevamente períodos de crecimiento y disminución. Sin embargo, desde 2024 se aprecia una tendencia general hacia niveles más bajos del índice. En conjunto, la presencia de numerosos picos y depresiones muestra que la actividad industrial presenta fluctuaciones a lo largo del tiempo, por lo que su comportamiento no puede describirse únicamente mediante una tendencia creciente o decreciente.
