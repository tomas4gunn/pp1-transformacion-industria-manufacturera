```python
columnas_numericas = df.columns.drop("indice_tiempo")

for columna in columnas_numericas:
    valores = pd.to_numeric(df[columna], errors="coerce")
    invalidos = df[columna].notna() & valores.isna()

    if invalidos.any():
        print(f"\n{columna}")
        print(df.loc[invalidos, columna])
```

```python
df = df.drop(columns=['ica_bienes_capital_partes_piezas',
             'ica_bienes_intermedios_combustibles_lubricantes',
             'ica_importaciones_combustibles_lubricantes',
             'ica_importaciones_piezas_accesorios_bienes_capital',
             'ica_importaciones_vehiculos_automotores_pasajeros',
             'ica_importaciones_resto',
             'ica_importaciones_bs_consumo_vehiculos_automotor_pasajeros'])  
```

```python
df
```

```python

# Seleccionar todas las columnas numéricas
columnas_numericas = df.select_dtypes(include="number").columns

# Recorrer cada columna numérica
for columna in columnas_numericas:
    serie = df[columna]
    
    # Calcular cuartiles e IQR
    Q1 = serie.quantile(0.25)
    Q3 = serie.quantile(0.75)
    IQR = Q3 - Q1

    # Calcular límites
    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR

    # Identificar valores atípicos
    atipicos = df[
        (df[columna] < limite_inferior) |
        (df[columna] > limite_superior)
    ]

    # Mostrar resultados
    print(f"\nColumna: {columna}")
    print(f"Cantidad de valores atípicos: {len(atipicos)}")
    print(atipicos[["indice_tiempo", columna]])
```

```python
df = df.rename(columns={'indice_tiempo':'fecha'})
```
